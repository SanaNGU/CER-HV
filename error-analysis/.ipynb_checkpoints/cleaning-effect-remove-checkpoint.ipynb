# 📘  CER-HV cleaning Error Impact Analysis Notebook  
### *CER-HV: CER Ranking with Human Verification*

This notebook extends error analysis performed in the  
**CER-HV (CER Ranking with Human Verification)** framework for the **train**, **val**,  
and **test** splits of the : 

- **[Muharaf dataset (Arabic - historic)](#muharaf)**.
- **[Ajami dataset (Hausa+Fulfulde - historic )](#ajami)**.
- **[PHTI dataset (Pashto - modern)](#phti)**.
- **[NUST dataset (Urdu - modern)](#nust)**.
- **[KHATT dataset (Arabic- modern)](#khatt)**.

The goal is to understand how different manually labeled error categories  
(e.g., *Segmentation error*, *Wrong transcription*, etc.) individually affect  
model performance.

---

## 🧮 1. Count Error Categories for Each Split

Use the following command to compute how many samples belong to each error  
category in a given split:

```python
%run error_counts.py <data>/hitl_labels_log_<split>.csv
```

## 🧮 2. See the effect of removing each  error category on CER for each eplit 

Use the following command to compute the effect of removing each  error category on CER for each split :
    
```python
%run   error_cleaning_effects.py \
  muharaf/muharaf_<split>_results_epoch_850_wih_imageid.csv \
  muharaf/hitl_labels_log_<split>.csv
```
## 3. to see the result after cleaning the train set pass --split train keyword. This will skip loading the HITL log and only compute/print CER+WER
