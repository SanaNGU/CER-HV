# 📘 KHATT Dataset — CRHV Error Analysis Notebook  
### *CRHV: CER Ranking with Human Verification*

This notebook analyzes results obtained from the  
**CRHV (CER Ranking with Human Verification)** framework across the **train**, **val**,  
and **test** splits of the **Muharaf dataset**.

---

## 🧮 1. Count Error Categories for Each Split

To compute how many samples belong to each error category, run:

```python
%run error_counts.py khatt/hitl_labels_log_test.csv
```

Change the path based on the dataset split you want to analyze:

- `khatt/hitl_labels_log_train.csv`
- `khatt/hitl_labels_log_val.csv`
- `khatt/hitl_labels_log_test.csv`

The script prints:

- The list of error categories found in the `Label` column  
- The number of samples per category  
- The total number of labeled samples (excluding empty labels and "Valid but hard")

---

## 🖼️ 2. Visualize Samples by Error Category

To visualize example images from each error category, run:

```python
%run visualize_hitl_by_label.py khatt/hitl_labels_log_test.csv --split test --max-per-group 5
```

Where:
- split selects the dataset split (train, val, test)
- max-per-group defines how many images to show per category

Examples:
```python
%run visualize_hitl_by_label.py khatt/hitl_labels_log_train.csv --split train --max-per-group 10
%run visualize_hitl_by_label.py khatt/hitl_labels_log_val.csv   --split val   --max-per-group 8
%run visualize_hitl_by_label.py khatt/hitl_labels_log_test.csv  --split test  --max-per-group 5

```
